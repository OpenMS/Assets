# **OpenMS-Tally-Banner-Script**

```python
import os
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.colors import LinearSegmentedColormap
from matplotlib.font_manager import FontProperties

# ==============================================================================
# 1. FONT CONFIGURATION (PUT YOUR PATH HERE)
# ==============================================================================
FONT_PATH = "../fonts/Montserrat/static/Montserrat-ExtraBold.ttf"

if os.path.exists(FONT_PATH):
    font_header = FontProperties(fname=FONT_PATH)
else:
    print(f"[WARNING] Font file not found at '{FONT_PATH}'. Falling back to system bold font.")
    font_header = FontProperties(family="sans-serif", weight="bold")

# ==============================================================================
# 2. BRAND DESIGN SPECS
# ==============================================================================
bg_color = '#231f20'  # Official Logo Black
brand_gradient = ['#3157e9', '#b503ff', '#ff03cb', '#ff7501', '#ffcb4f']  # Core palette

# ==============================================================================
# 3. CANVAS & PLOT SETUP (1128 x 191)
# ==============================================================================
dpi = 1000
fig, ax = plt.subplots(figsize=(11.28, 1.91), dpi=dpi)
fig.patch.set_facecolor(bg_color)
ax.set_facecolor(bg_color)

ax.axis('off')
fig.subplots_adjust(left=0, right=1, top=1, bottom=0)

cmap = LinearSegmentedColormap.from_list('openms_gradient', brand_gradient, N=256)

# ==============================================================================
# 4. GENERATE DATA (Mass Spec Spectrum Peaks)
# ==============================================================================
x = np.linspace(0, 10, 1200)
y = np.random.normal(0.012, 0.002, len(x))

# Shifted peak center points up to start at 6.0 and run through 9.7
peaks = [
    (6.0, 0.35, 0.06), (6.3, 0.18, 0.04), (6.7, 0.45, 0.05),
    (7.1, 0.12, 0.03), (7.5, 0.88, 0.04), (7.9, 0.42, 0.05),
    (8.4, 0.55, 0.04), (8.9, 0.72, 0.04), (9.3, 0.28, 0.06),
    (9.7, 0.15, 0.03)
]

for pos, height, width in peaks:
    y += height * np.exp(-((x - pos) / width) ** 2)
y = np.clip(y, 0, None)

start_gradient_x = 5.6
end_gradient_x = 9.8

# Loop modified to run across all data points from x=0 to x=10
for i in range(len(x) - 1):
    norm_val = (x[i] - start_gradient_x) / (end_gradient_x - start_gradient_x)
    color = cmap(np.clip(norm_val, 0, 1))  # Anything < 5.6 safely defaults to solid blue
    ax.plot(x[i:i+2], y[i:i+2], color=color, lw=2.5, solid_capstyle='round')
    ax.fill_between(x[i:i+2], 0, y[i:i+2], color=color, alpha=0.05)

# ==============================================================================
# 5. TYPOGRAPHY INJECTION (Precise Layout, Indent & Line Spacing)
# ==============================================================================
text_color = '#ffffff'
text_x = 3.5  # Renamed from 'x' to avoid wiping out the data vector

# Line 1
ax.text(
    x=text_x,  
    y=0.64,  
    s="open science.",  
    color=text_color,  
    fontproperties=font_header,  
    fontsize=26,  
    va='center',  
    ha='left'
)

# Line 2
ax.text(
    x=text_x,  
    y=0.30,  
    s="clear results.",  
    color=text_color,  
    fontproperties=font_header,  
    fontsize=26,  
    va='center',  
    ha='left'
)

# ==============================================================================
# 6. EXPORT (With Height Adjustments for Clean Padding)
# ==============================================================================
ax.set_xlim(0, 10)
ax.set_ylim(-0.15, 1.15)  # Keeps the zero-intensity line safely above the bottom edge

output_filename = '../banners/openms_tally_company_banner.png'
plt.savefig(output_filename, facecolor=fig.get_facecolor(), edgecolor='none', dpi=dpi)
plt.close()

print(f"Success! Company banner saved as '{output_filename}' (1128x191 px).")
```
